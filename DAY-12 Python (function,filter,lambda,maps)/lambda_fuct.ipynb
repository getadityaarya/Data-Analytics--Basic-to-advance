# Lambda Function

A lambda function in Python is a small anonymous function that is defined using the lambda keyword. It can take any number of arguments but contains only a single expression, which is evaluated and returned automatically.

```python
lambda arguments : expression
```

```python
# Example 

square = lambda x: x * x

print(square(5))
```

```python
# Find maximum

maximum = lambda a, b: a if a > b else b

print(maximum(15, 8))
```

```python
# Find even 
even = lambda x:"Even" if x%2 == 0 else "Odd"

print(even(6))
```

```python
# Convert Celsius to Fahrenheit
convert = lambda c: (c * 9/5) + 32

print(convert(25))
```
