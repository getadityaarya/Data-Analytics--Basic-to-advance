# filter() Function

The filter() function is used to select elements from an iterable based on a condition. It keeps only those elements for which the function returns True

```python
# filter(function, iterable)
```

```python
# Even numbers

nums = [1, 2, 3, 4, 5, 6]

result = list(filter(lambda x: x % 2 == 0, nums))

print(result)
```

```python
# Greater than 50 

marks = [30,60,80,40,90]

result = list(filter(lambda x: x>50, marks))
print(result)
```

```python
# Remove empty strings

words = ["hello","","python","","AI"]

result = list(filter(lambda x: x!="", words))

print(result)
```

```python
# Find vowels
letters = ['a','b','e','c','i']

isVowel = list(filter(lambda x: x in 'aeiou', letters))
print(isVowel)
```
