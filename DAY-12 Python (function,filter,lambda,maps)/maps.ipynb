# map() Function

The map() function is used to apply a function to every element of an iterable (like list, tuple) and returns a new iterable containing the results.

```python
# Syntax -> map(function, iterable)
```

```python
# Example - Square numbers

nums = [1, 2, 3, 4]

result = list(map(lambda x: x * x, nums))

print(result)
```

```python
#Convert strings to uppercase

names = ["aditya", "rahul", "amit"]

answer= list(map(lambda x: x.upper(),names))
print(answer)
```

```python
# Add two lists

a = [1,2,3]
b = [4,5,6]

addition = list(map(lambda x,y:x+y, a,b))
print(addition)
```

```python
# Increase salary by 10 percent

salary = [30000,40000,50000]

increment  = list(map(lambda x: x+ (x*1/10),salary))
print(increment)
```
