## For loop
Used when you want to iterate over a sequence (list, string, range, etc.) or when you know how many times to repeat.

```python
for i in range(5):
    print(i)
```

# While loop
Used when you want to repeat something as long as a condition is true.

```python
i = 0

while i < 5:
    print(i)
    i += 1
```

# Nested loop

A loop inside another loop.

```python
for i in range(3):
    for j in range(2):
        print(i, j)
```

### For loop inside a while loop

```python
i = 1

while i <= 3:
    for j in range(1, 4):
        print(i, j)
    i += 1
```

# Infinite loop

A loop that never stops because its condition always remains true.

```python
while True:
    print("Running")
```

# Loop control statements (used with loops)

### Break → stops the loop

```python
for i in range(5):
    if i == 3:
        break
    print(i)
```

### Continue → skips current iteration

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
    
```

### Pass → placeholder, does nothing

```python
for i in range(5):
    pass
```
