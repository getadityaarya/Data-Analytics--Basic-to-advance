# Set in Python

A set is a collection of unique (non-duplicate) elements.
Sets are unordered and mutable.

```python
numbers = {1, 2, 3, 4, 5}

print(numbers)
```

#### Important properties of Set:

No duplicate values

Unordered (no indexing)

Mutable (can add/remove items)

Supports mathematical operations (union, intersection, etc.)

```python
# Duplicate values are removed automatically:

numbers = {1, 2, 2, 3, 3, 4}

print(numbers)
```

# Creating a Set


```python
# Using {}
my_set = {"apple", "banana", "mango"}
print(my_set)
```

```python
#Using set()
my_set = set([1, 2, 3])
print(my_set)
```

Adding Elements - add()

```python
s = {1, 2, 3}

s.add(4)

print(s)
```

Upadate - Add more one element in set at once. 

```python
numbers = {1, 2, 3}

numbers.update([4, 5, 6])
print(numbers)  

numbers.update({6, 7, 8})
print(numbers)  
```

Removing Elements - remove()

```python
s = {1, 2, 3}

s.remove(2)

print(s)
```

Discard() - Does not give error if item doesn't exist.

```python
s.discard(10)
```

# Set Operations

1. Union ( | ) - Combines both sets.

```python
A = {1,2,3}
B = {3,4,5}

print(A | B)
```

2. Intersection (&) - Common elements.

```python
print(A & B)
```

3. Difference (-) - Elements only in first set.

```python
print(A - B)
```

4. Symmetric Difference (^) - Elements that are not common.

```python
print(A ^ B)
```

#### Set Comprehension

Similar to list comprehension.

```python
squares = {x*x for x in range(5)}

print(squares)
```

# Questions

```python
# Counting unique words in text.

text = "Goodmorning Friends! My name is Aditya Arya. Today is a worderful day.Fun fact of the day is - Sun helps us to see,  without its help we can not see anything."
words = text.split()
```

```python
# Convert list of words to set to get all unique words.
unqiue_words = set(words)
print(unqiue_words)
print(len(unqiue_words))
```

```python
# Find unique characters in a string
text = "programming"
unique_chars = set(text)

print(unique_chars)
# Output: {'p', 'r', 'o', 'g', 'a', 'm', 'i', 'n'}
```

```python
#Alphabetical Order:
unique_sorted = sorted(set("programming"))
print(unique_sorted)
```

```python
#Check if two strings are anagrams using sets
a = "listen"
b = "silent"
print( set(a) == set(b))
```
