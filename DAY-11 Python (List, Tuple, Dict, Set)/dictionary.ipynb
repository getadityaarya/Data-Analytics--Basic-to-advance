# Dictionary 

A built-in data structure that stores data in key: value pairs, where each unique key maps to a specific value

Create

```python
# Create dictionary
student = {
    "name": "Aditya",
    "age": 22,
    "course": "AI"
}

print(student)
```

```python
# Access values
print(student["name"])
print(student.get("age"))
```

```python
# Add and update values
student["city"] = "Pune"   # add
student["age"] = 23       # update

print(student)
```

```python
# Delete values
student.pop("course")

print(student)
```

```python
del student["age"]
print(student)
```

### Loop through dictionary

```python
#Keys
for key in student:
    print(key)
```

```python
#Value
for value in student.values():
    print(value)
```

```python
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
# Check key exists
if "name" in student:
    print("Exists")
```

```python
# Dictionary length
print(len(student))
```

#### Nested dictionary

```python
students = {
    "student1": {
        "name": "Aditya",
        "marks": 90
    },
    "student2": {
        "name": "Rahul",
        "marks": 85
    }
}

print(students["student1"]["marks"])
```

#### Dictionary comprehension

```python
square = {x: x*x for x in range(5)}

print(square)
```

# Questions

```python
# Frequency counter (very important interview)
word = "banana"

freq = {}

for ch in word:
    freq[ch] = freq.get(ch, 0) + 1

print(freq)
```

#### Sort dictionary

```python
# By keys:

d = {"c":3, "a":1, "b":2}

print(dict(sorted(d.items())))
```

```python
# By values:

print(dict(sorted(d.items(), key=lambda x:x[1])))
```

#### Merge dictionaries


```python
a = {"name":"Aditya"}
b = {"age":22}

c = a | b

print(c)
```
