### List in Python

A list is a collection of items stored in a single variable. Lists are ordered, mutable (changeable), and allow duplicates.

```python
numbers = [1, 2, 3, 4, 5]

print(numbers)
```

Access elements:

```python
numbers = [10, 20, 30]

print(numbers[0])
```

Add items:

```python
numbers.append(40)
print(numbers)
```

# List Comprehension

List comprehension is a shorter way to create a list using a single line of code

```python
# Normal way:

squares = []

for i in range(5):
    squares.append(i*i)

print(squares)
```

```python
# Using list comprehension:

squares = [i*i for i in range(5)]

print(squares)
```

##### Syntax: [expression for item in iterable if condition]

```python
even = [i for i in range(10) if i % 2 == 0]

print(even)
```

##### if-else in list comprehension:

```python
result = ["Even" if i%2==0 else "Odd" for i in range(5)]

print(result)
```
