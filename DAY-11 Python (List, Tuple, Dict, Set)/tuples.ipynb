# Tuples

A tuple is an ordered, immutable collection of elements in Python.

### Main Properties of Tuple

Ordered -   Elements maintain their position.

Immutable -   Once created, elements cannot be changed.

```python
# Create a tuple
numbers = (1, 2, 3, 4)

print(numbers)
```

```python
# Single element tuple
t = (2,)
print(type(t))
```

```python
# Access tuple elements
t = ("Aditya", 22, "AI")

print(t[0])
print(t[1])
```

```python
# Indexing & slicing
t = (10, 20, 30, 40, 50)

print(t[1:4])
```

```python
# Tuple is immutable - You cannot change:

t = (1,2,3)

t[0] = 10
```

```python
# Loop through tuple
t = ("a","b","c")

for x in t:
    print(x)
```

```python
# Check element exists
t = (10,20,30)

print(20 in t)
```

### Tuple methods

```python
# count()
t = (1,2,2,3,2)

print(t.count(2))
```

```python
# index()
t = ("a","b","c")

print(t.index("b"))
```

### Packing and unpacking 

```python
# Packing:

student = "Aditya", 22, "AI"

print(student)
```

```python
# Unpacking:

name, age, branch = student

print(name)
print(age)
```

```python
# Swap variables using tuple
a = 10
b = 20

a, b = b, a

print(a,b)
```

```python
# Nested tuple
data = (
    ("Aditya",22),
    ("Rahul",23)
)

print(data[0][1])
```

```python
# Tuple inside dictionary (common interview)
student = {
    "marks": (90, 85, 95)
}

print(student["marks"])
```

```python
# List to tuple:

lst = [1,2,3]

t = tuple(lst)

print(t)
```

### Questions

```python
# Reverse a tuple
t = (1,2,3,4)

print(t[::-1])
```

```python
# Find maximum value
t = (10,50,20)

print(max(t))
```

```python
# Remove duplicates
t = (1,2,2,3,3)

result = tuple(set(t))

print(result)
```

```python
# Count frequency
t = (1,2,2,3,2)

for x in set(t):
    print(x, t.count(x))
```
